```python
import requests
from bs4 import BeautifulSoup
import nltk
from collections import Counter

nltk.download('stopwords')
```

```python
## Step 1: Request the Moby Dick HTML file
url = 'https://s3.amazonaws.com/assets.datacamp.com/production/project_147/datasets/2701-h.htm'

r = requests.get(url)
r.encoding = 'utf-8'
html = r.text
print(html[0:2000])
```

```python
## Step 2: Extract text with BeautifulSoup
html_soup = BeautifulSoup(html, 'html.parser')
moby_text = html_soup.get_text()
print(moby_text[32000:34000])
```

```python
## Step 3: Tokenize the text
tokenizer = nltk.tokenize.RegexpTokenizer(r'\w+')
tokens = tokenizer.tokenize(moby_text)
print(tokens[:8])
```

```python
## Step 4: Lowercase and remove stop words
words = [token.lower() for token in tokens]
stop_words = nltk.corpus.stopwords.words('english')
words_no_stop = [word for word in words if word not in stop_words]
print(words_no_stop[:8])
```

```python
## Step 5: Find the ten most common words
count = Counter(words_no_stop)
top_ten = count.most_common(10)
print(top_ten)
```
