In this workspace, you'll scrape the novel Moby Dick from the website [Project Gutenberg](https://www.gutenberg.org/) (which contains a large corpus of books) using the Python `requests` package. You'll extract words from this web data using `BeautifulSoup` before analyzing the distribution of words using the Natural Language ToolKit (`nltk`) and `Counter`.

The Data Science pipeline you'll build in this workspace can be used to visualize the word frequency distributions of any novel you can find on Project Gutenberg.

```python
# Import and download packages
import requests
from bs4 import BeautifulSoup
import nltk
from collections import Counter
nltk.download('stopwords')

# Requesting and encoding from the link provided
pedido = requests.get('https://s3.amazonaws.com/assets.datacamp.com/production/project_147/datasets/2701-h.htm')
pedido.encoding = 'utf-8'
text = pedido.text

# Cleaning the text with beautiful soup and then extracting the Moby Dick reading.
soupify = BeautifulSoup(text, 'html5lib')
reading = soupify.get_text()

#tokenizer set to keep alphanumeric text, as demanded by the rubric.
tokenizer = nltk.tokenize.RegexpTokenizer('\w+')
tokens = tokenizer.tokenize(reading)
unnecessary = nltk.corpus.stopwords.words('english') #unnecessary will store out stop words list.
words_no_stop = [] #this array will store our list of words used in the reading

#this for loop will append the words in our word list that are not in the unnecessary variable, this will only store lowercase tokens that are not stop words
for word in tokens: 
    if (word.lower() not in unnecessary):
        words_no_stop.append(word.lower()) #set all words in our list in lowercase.

#A counter Object is initialized, the ten "most_common()" words used are stored in top_tem to then be printed.
top_ten = Counter(words_no_stop).most_common(10)

# Print the result
print(top_ten)
```
